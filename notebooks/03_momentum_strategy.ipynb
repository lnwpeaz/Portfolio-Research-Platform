```python
import pandas as pd
import numpy as np

prices = pd.read_parquet("../data/raw/us_stock.parquet")

close = prices["Close"]
returns = close.pct_change(fill_method=None)
```

```python
momentum = close.pct_change(60)
```

```python
today = momentum.iloc[-1]

today.sort_values(
    ascending=False
).head(5)
```

```python
top5 = today.sort_values(
    ascending=False
).head(5)

weights = pd.Series(
    1/len(top5),
    index=top5.index
)

weights
```

```python
portfolio_return = (
    returns[top5.index].iloc[-1] *
    weights
).sum()

portfolio_return
```
